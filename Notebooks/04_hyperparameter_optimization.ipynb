Este cuaderno implementa de manera rigurosa la optimización de hiperparámetros mediante GridSearchCV para el árbol de decisión, buscando controlar la complejidad del modelo y mitigar el sobreajuste.

```python
# Celda 1: Importación de herramientas de optimización
import pandas as pd
from sklearn.model_selection import GridSearchCV
from sklearn.tree import DecisionTreeClassifier
import joblib

# Celda 2: Carga de particiones de datos estandarizadas
X_train_scaled, X_test_scaled, y_train, y_test = joblib.load('data_partitions.pkl')

# Celda 3: Definición del espacio de hiperparámetros (Grilla)
dt_model = DecisionTreeClassifier(random_state=42)

param_grid = {
    'criterion': ['gini', 'entropy'],
    'max_depth': [3, 5, 8, 12, None],
    'min_samples_split': [2, 5, 10, 15],
    'min_samples_leaf': [1, 2, 4, 6]
}

# Celda 4: Configuración y ejecución de GridSearchCV con Validación Cruzada (CV=5)
grid_search = GridSearchCV(
    estimator=dt_model,
    param_grid=param_grid,
    cv=5,
    scoring='f1_macro', # Evaluamos optimizando un balance macro de las clases
    n_jobs=-1,
    verbose=1
)

grid_search.fit(X_train_scaled, y_train)

# Celda 5: Resultados de la optimización
print("Mejores hiperparámetros encontrados:")
print(grid_search.best_params_)
print(f"\nMejor puntuación F1-Macro en Validación Cruzada: {grid_search.best_score_:.4f}")

# Celda 6: Exportación del mejor modelo optimizado obtenido
best_dt_optimized = grid_search.best_estimator_
joblib.dump(best_dt_optimized, 'model_dt_optimized.pkl')
print("Modelo optimizado guardado como 'model_dt_optimized.pkl'.")
```
